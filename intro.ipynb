## Connecting Colab to GitHub

1.  **Open a notebook in Colab.**
2.  Go to **File > Save a copy in GitHub**.
3.  A pop-up window will appear asking you to authorize Colaboratory to access your GitHub repositories. Click **Authorize GoogleColab**.
4.  Follow the prompts to sign in to GitHub if you aren't already logged in.
5.  After authorization, you can choose the GitHub repository and filename for your notebook. You can also add a commit message.
6.  Click **OK** to save the notebook to your GitHub repository.

**To open a notebook from GitHub:**

1.  Go to **File > Open notebook**.
2.  Switch to the **GitHub** tab.
3.  You can then select a repository and open a notebook from there.
